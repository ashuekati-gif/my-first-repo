```python

import numpy as np
from scipy.stats import chisquare, chi2_contingency

# Part 1: Goodness-of-Fit Test
observed = np.array([8, 12, 9, 11, 10, 10])
expected = np.array([10, 10, 10, 10, 10, 10])

chi_stat, p_value = chisquare(observed, f_exp=expected)
df = len(observed) - 1

print("PART 1: Goodness-of-Fit Test")
print("Chi-Square Statistic:", round(chi_stat, 4))
print("Degrees of Freedom:", df)
print("p-value:", round(p_value, 4))

if p_value < 0.05:
    print("Conclusion: Reject the null hypothesis. The die is not fair.")
else:
    print("Conclusion: Fail to reject the null hypothesis. The die may be fair.")

# Part 2: Test of Independence
table = np.array([
    [30, 40, 30],
    [40, 30, 30]
])

chi_stat, p_value, df, expected_freq = chi2_contingency(
    table, correction=False
)

print("\nPART 2: Test of Independence")
print("Chi-Square Statistic:", round(chi_stat, 4))
print("Degrees of Freedom:", df)
print("p-value:", round(p_value, 4))
print("Expected Frequencies:")
print(expected_freq)

if p_value < 0.05:
    print("Conclusion: Reject the null hypothesis. Gender and product preference are associated.")
else:
    print("Conclusion: Fail to reject the null hypothesis. No significant association between gender and product preference.")
```

Observed Frequencies: [8, 12, 9, 11, 10, 10]
Expected Frequencies: [10, 10, 10, 10, 10, 10]
